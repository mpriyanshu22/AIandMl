```python
import seaborn as sns
tips=sns.load_dataset('tips')
```

```python
tips
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill',y='tip',data=tips)
plt.title("scatter plot of total bill is tip")
plt.show()
```

```python
sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("line of size vs tital bill")
plt.show()
```

```python
#categorial variable plot
tips
#sns.barplot(x='day',y='totol_bill')
```

```python
tips.head()
```

```python
sns.barplot(x='day',y='total_bill',data=tips)
plt.show()
```

```python
#box plot 
sns.boxplot(x='day',y='total_bill',data=tips)
plt.title("box plot of total bill by day")
plt.show()
```

```python
#voilinplot
sns.violinplot(x='day',y='total_bill',data=tips)
```

```python
sns.histplot(tips['total_bill'],bins=5,kde=True)
```

```python
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
sns.pairplot(tips)
```

```python
#heatmap
corr=tips[['total_bill','tip','size']].corr()
corr
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd

data_file = pd.read_csv(
    'tata_data.csv',
    skiprows=2,
    names=['Date', 'Close', 'High', 'Low', 'Open', 'Volume']
)

data_file.head()
data_file.drop(0,inplace=True)
```

```python
data_file
```

```python
data_file.head()
```

```python
data_file.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
sns.barplot(x='Date',y='Close',data=data_file,estimator=sum)
plt.title("Date vs Close")
plt.xlabel('Date')
plt.ylabel('Close')
plt.show()
```
