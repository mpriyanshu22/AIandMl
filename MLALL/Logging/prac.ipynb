```python
import logging
# logging.basicConfig(level=logging.DEBUG)
```

```python
##log messages with different security levels
logging.debug("this a debug message")
logging.info("this is an infomessage")
logging.warning("this is s warning message")
logging.error("this is a error message")
logging.critical("this is message for critical error")
```

```python
##configuring logging
logging.basicConfig(
level=logging.DEBUG,
format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
datefmt='%Y-%m-%d %H:%M:%S',
filename='app.log',
filemode='w',
)
```
