```python
import threading
import time
def print_numbers():
    for i in range(5):
        print(f"Number:{i}")

def print_letter():
    for letter in "abcde":
        print(f"Letter:{letter}")         
```

```python
starttime=time.time()
print_numbers()
print_letter()
endtime=time.time()
finishing_time=starttime-endtime
print(finishing_time)
```

```python
##multithreading
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_letter)
starttime=time.time()
t1.start()
t2.start()
t1.join()
t2.join()
```

```python
import multiprocessing
import time

def sq_num():
    for i in range(5):
        print(f"square:{i*i}", flush=True)

def cube_num():
    for i in range(5):
        print(f"cube:{i*i*i}", flush=True)

if __name__ == "__main__":
    p1 = multiprocessing.Process(target=sq_num)
    p2 = multiprocessing.Process(target=cube_num)

    t = time.time()
    p1.start()
    p2.start()
    p1.join()
    p2.join()

    finished_time = time.time() - t
    print(finished_time)
```

```python
from concurrent.futures import ThreadPoolExecutor
import time
def print_number(number):
    # time.sleep(1)
    return f"Number:{number}"
numbers=[1,2,3,4,5]

def print_letter(letter):
    # time.sleep(1)
    return f"letter:{letter}"   
with ThreadPoolExecutor(max_workers=3) as executor:
    results_num=executor.map(print_number,numbers)
    results_letter=executor.map(print_letter,"abcde")
for result in results_num:
    print(result)    
for result in results_letter:
    print(result)      
```
