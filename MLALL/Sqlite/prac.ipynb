```python
import sqlite3
##connect to sqlite
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
cursor
```

```python
cursor.execute('''
      Create Table if not Exists employees(
              id Integer Primary Key,
               name Text Not Null,
               age Integer,
               department Text
               )
''')
connection.commit()
```

```python
cursor.execute('''
     select * from employees
''')
```

```python
cursor.execute('''
     Insert into employees(name,age,department)
     values('Anki',22,'it')
               ''')
cursor.execute('''
     Insert into employees(name,age,department)
     values('munkey',22,'it')
               ''')
cursor.execute('''
     Insert into employees(name,age,department)
     values('junky',22,'it')
               ''')
cursor.execute('''
     Insert into employees(name,age,department)
     values('funki',22,'it')
               ''')
```

```python
connection.commit()
```

```python
##project
connection=sqlite3.connect('sales.db')
cursor=connection.cursor()
```

```python
cursor.execute('''
create table if not exists sales_data(
             id integer primary key,
               date text not null,
               product text not null,
               sales integer,
               region text
               )
''')

cursor.execute('select * from sales_data')
```

```python
sales_val=[
    ('24-5-26','product1',100,'north'),
    ('25-5-26','product2',100,'west'),
    ('26-5-26','product3',100,'east'),
    ('27-5-26','product4',100,'south'),
    ('28-5-26','product5',100,'north')
]
cursor.executemany('''
insert into sales_data(date,product,sales,region)
               values(?,?,?,?)
''',sales_val)
cursor.execute('select * from sales_data')
rows=cursor.fetchall()
for row in rows:
    print(row)
```

```python
connection.close()
```
