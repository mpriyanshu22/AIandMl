```python
import pandas as pd

data = {
    "Department": ["IT", "IT", "HR", "HR", "Finance", "IT"],
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Salary": [50000, 60000, 40000, 45000, 70000, 55000],
    "Bonus": [5000, 6000, 3000, 3500, 7000, 5200]
}

df = pd.DataFrame(data)
print(df)
```

```python
grouped_mean=df.groupby('Department')['Salary'].mean()
print(grouped_mean)
```

```python
df
```

```python
grouped_sum=df.groupby(['Department','Employee'])['Salary'].sum()
print(grouped_sum)
```

```python
grouped_agg=df.groupby('Department')['Salary'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
df1=pd.DataFrame({'key':['A','B','C'],
                  'Value1':[1,2,3]
                  })
df2=pd.DataFrame({'key':['A','B','D'],
                  'Value2':[4,5,6]
                  })
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="key",how="inner")
```

```python
pd.merge(df1,df2,on="key",how="outer")
```

```python
pd.merge(df1,df2,on="key",how="left")
```

```python
pd.merge(df1,df2,on="key",how="right")
```
